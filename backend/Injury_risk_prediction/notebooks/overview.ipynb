- wake_ups: The total number of times you wake up during the night after initially falling asleep.
- resting_heart_rate (RHR): Your heart rate measured while you are at complete rest, typically during your deepest sleep.
- rhr_baseline: The rolling average of your resting heart rate over a longer period (e.g., several weeks).
- respiratory_rate: The number of breaths you take per minute during sleep. 
- activity_strain: A score on a 0 to 21 scale that measures the amount of strain placed on heart and circulatory system during activity
- activity_calories: The estimated number of calories burned specifically during a recorded activity
- HR Zones : Is a 5-level intensity scale for heart. Instead of just giving you one average heart rate, it breaks your workout down into "buckets" of time to show how hard you were actually pushing.
- day_strain: This is your total cardiovascular load for the entire 24-hour period, measured on a logarithmic scale from 0 to 21. It includes not just your workouts, but also the stress of daily life
- recovery_score: A percentage from 0% to 100% that tells you how prepared your body is for the day.
- hrv: measures the variation in time between heart beats (e.g., one beat-to-beat interval might be seconds, the next).

##  Feature Engineering

### 1.3. Creating the Injury Label (Injury Risk Proxy)

```python
def create_injury_label(row):

    if row["day_strain"] > 15 and row["recovery_score"] < 40:
        return 1
    
    if row["recovery_score"] < 20 and row["hrv"] < 20:
        return 1
    
    if row["day_strain"] > 19:
        return 1
    return 0 


df['injury_label'] = df.apply(create_injury_label, axis=1)  # apply create_injury_label across each row in df
df  
```

```python
df.isna().sum()
```

```python

# sort data 
df = df.sort_values(['user_id', "date"])



# Lag Features: Create features from the previous days to capture trends.

windows = [7, 14, 28]
metrics = ['day_strain', 'hrv', 'sleep_hours', 'resting_heart_rate']

for window in windows:
    for metric in metrics:

        df[f"{metric}_roll_mean_{window}"] = df.groupby('user_id')[metric].transform(lambda x: x.rolling(window=window, min_periods=1).mean() )
        df[f"{metric}_roll_std_{window}"] = df.groupby('user_id')[metric].transform(lambda x: x.rolling(window=window, min_periods=1).std() )


# Rate of Change : Calculate the day-over-day change for key metrics.
for metric in metrics:
    df[f"{metric}_roc"] = df.groupby('user_id')[metric].pct_change()  # percentage rate of change
    df[f"{metric}_roc"] = df[f"{metric}_roc"].replace([np.inf, -np.inf, np.nan], 0)

df     
```

**Acute:Chronic Workload Ratio (ACWR)**:  
calculate the ratio of acute (7-day) to chronic (28-day) workload.

### Composite Strain Index

```python
# Normalize each metric to 0-1 scale
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
df['strain_norm'] = scaler.fit_transform(df[['day_strain']])
df['hrv_inverse_norm'] = 1 - scaler.fit_transform(df[['hrv']])  # low HRV = high load
df['sleep_deficit_norm'] = 1 - scaler.fit_transform(df[['sleep_hours']]) / 8

# Composite load (weights based on domain knowledge)
df['daily_load'] = (0.6 * df['strain_norm'] + 
                     0.3 * df['hrv_inverse_norm'] + 
                     0.1 * df['sleep_deficit_norm'])

# Then compute ACWR on 'daily_load'
acute = df.groupby('user_id')['daily_load'].transform(lambda x: x.rolling(7, min_periods=1).mean())
chronic = df.groupby('user_id')['daily_load'].transform(lambda x: x.rolling(28, min_periods=1).mean())
df['acwr_composite'] = acute / (chronic + 1e-6)

df
```

```python

df['date'] = pd.to_datetime(df['date'])

df['day_of_week'] = df['date'].dt.dayofweek  
df['month'] = df['date'].dt.month
df['day_of_month'] = df['date'].dt.day  
df['weekend'] = (df['date'].dt.dayofweek >= 5).astype(int)


# Cyclical encoding for day_of_week and month (important for tree models). December (11) and January (0) will have coordinates very close to each other on the circle. 
df['day_of_week_sin'] = np.sin(2 * np.pi * df['day_of_week'] / 7)
df['day_of_week_cos'] = np.cos(2 * np.pi * df['day_of_week'] / 7)
df['month_sin'] = np.sin(2 * np.pi * df['month'] / 12)
df['month_cos'] = np.cos(2 * np.pi * df['month'] / 12)


# Then drop the original 'date' column before training
df = df.drop(columns=['date'])
df 
```

### categorical feature

```python
# Replace 'N/A' with a new category, 'Unknown'

df['workout_time_of_day'] = df['workout_time_of_day'].fillna('Unknown')

```

```python
# cyclic encoding of 'workout_time_of_day'

# First, map your categories to a numeric hour
time_map = {'Morning': 8, 'Afternoon': 14, 'Evening': 19, 'Unknown': 12}
df['hour_of_day'] = df['workout_time_of_day'].map(time_map)

# Apply cyclical encoding (sine/cosine transformation)
def encode_cyclical(data, col, max_val):
    data[f'{col}_sin'] = np.sin(2 * np.pi * data[col] / max_val)
    data[f'{col}_cos'] = np.cos(2 * np.pi * data[col] / max_val)
    return data

# Encode the hour (max_val is 24)
df = encode_cyclical(df, 'hour_of_day', 24)

# drop the intermediate 'hour_of_day' column
df = df.drop(columns=['hour_of_day', 'workout_time_of_day'])
df 
```

##  Model Training 

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.1, random_state=42, stratify=y)

neg_count = (y_train == 0).sum()
pos_count = (y_train == 1).sum()

scale_pos_weight = neg_count / pos_count

print(f"Negative: {neg_count}, Positive: {pos_count}, Ratio: {scale_pos_weight:.2f}") 
```

```python
import optuna
from sklearn.model_selection import GroupKFold, cross_val_score
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier


def cv_score(model, X, y, cv_folds=5):
    
    gkf = GroupKFold(n_splits=cv_folds)
    groups = X['user_id']
    X = X.drop(columns=['user_id'])
    scores = cross_val_score(estimator=model, X=X, y=y, cv=gkf, groups=groups, scoring='neg_root_mean_squared_error')  
    return -scores.mean()


def objective_xgb(trial, X, y, cv_folds):
    params = {
        'n_estimators': trial.suggest_int('n_estimators', 50, 350, step=50),
        'max_depth': trial.suggest_int('max_depth', 1, 15),
        'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.3, log=True),
        'subsample': trial.suggest_float('subsample', 0.5, 1.0),
        'colsample_bytree': trial.suggest_float('colsample_bytree', 0.5, 1.0),
        'min_child_weight': trial.suggest_int('min_child_weight', 1, 15),
        'gamma': trial.suggest_float('gamma', 0, 5),
        'reg_alpha': trial.suggest_float('reg_alpha', 1e-8, 10.0, log=True),
        'reg_lambda': trial.suggest_float('reg_lambda', 1e-3, 10.0, log=True),
        "tree_method": "hist",  
        "device": "cuda",
    }
    
    model = XGBClassifier(**params, random_state=42, verbosity=0, scale_pos_weight=scale_pos_weight)
    score = cross_val_score(model, X, y, cv_folds=cv_folds)

    return score.mean() 


def objective_lgbm(trial, X, y, cv_folds):
    params = {
        'n_estimators': trial.suggest_int('n_estimators', 50, 350, step=50),
        'max_depth': trial.suggest_int('max_depth', 1, 10),
        'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.3, log=True),
        'bagging_fraction': trial.suggest_float('bagging_fraction', 0.5, 1.0),
        'feature_fraction': trial.suggest_float('feature_fraction', 0.5, 1.0),
        'min_child_samples': trial.suggest_int('min_child_samples', 1, 100),
        'reg_alpha': trial.suggest_float('reg_alpha', 1e-8, 10.0, log=True),
        'reg_lambda': trial.suggest_float('reg_lambda', 1e-4, 10.0, log=True),
        'num_leaves': trial.suggest_int('num_leaves', 10, 100),
        'min_split_gain': trial.suggest_float('min_split_gain', 0, 1),
        "device": "gpu"            
    }

    model = LGBMClassifier(**params, random_state=42, verbose=-1, scale_pos_weight=scale_pos_weight)
    score = cv_score(model, X, y, cv_folds)
    return score.mean()
```

```python
def tune_with_optuna(X_train, y_train, model_name, n_trials=50):

    X = X_train
    y = y_train
    
    objectives ={
        "XGBoost": objective_xgb,
        "LightGBM": objective_lgbm
    }

    study = optuna.create_study(direction='minimize', sampler=optuna.samplers.TPESampler(seed=42))

    study.optimize(lambda trail: objectives[model_name](trial=trail, X=X, y=y, cv_folds=5), 
                   n_trials=n_trials, show_progress_bar=True, n_jobs=-1)

    best_params = study.best_params
    best_rmse = study.best_value
    
    print(f"Best params: {best_params} -- Best rmse: {best_rmse}")

tune_with_optuna(X_train, y_train, model_name="LightGBM")    
```
